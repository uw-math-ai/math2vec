```python
from datasets import load_dataset
from sklearn.model_selection import train_test_split

ds = load_dataset("FrenzyMath/mathlib_informal_v4.19.0")

data = ds["train"]

# Split off 80% train, 20% temp
train_test = data.train_test_split(test_size=0.2, seed=1)
train_ds = train_test["train"]    

# Split the 20% temp into 10% val and 10% test
val_test = train_test["test"].train_test_split(test_size=0.5, seed=42)
val_ds   = val_test["train"]       
test_ds  = val_test["test"]        

print(f"Train: {len(train_ds):,} rows")   
print(f"Val:   {len(val_ds):,} rows")     
print(f"Test:  {len(test_ds):,} rows")

train_ds.to_json("train.jsonl")
val_ds.to_json("val.jsonl")
test_ds.to_json("test.jsonl")
```
